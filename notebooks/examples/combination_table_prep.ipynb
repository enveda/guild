```python
from copy import deepcopy

import pandas as pd
from tqdm import tqdm
```

```python
gpcr_df = pd.read_csv(
    "../../guild/support/publication_data/gpcrdb_protein_data_one.csv", header=0
)
gpcr_df_samples = gpcr_df.head(5)  # just 5 proteins

ligands_df = pd.read_csv(
    "../../guild/support/publication_data/ligands_sampled_1000.tsv", sep="\t"
)
ligands_df_samples = ligands_df.sample(5)
```

## Generate the input table from ligand and protein file

```python
input_table_rows = []

downloaded_pdbs_folder = "notebooks/data/pdbs"

for _idx, row in tqdm(
    gpcr_df_samples.iterrows(),
    total=len(gpcr_df_samples),
    desc="Processing GPCR proteins",
):
    protein_row = {
        "protein_id": row["pdb_id"],
        "protein_chain": row["Chain"],
        "protein_path": f"{downloaded_pdbs_folder}/{row['pdb_id']}.pdb",
        "original_ligand": row["Ligand"],
        "original_ligand_chain": row["Chain"],
        "protein_config_id": f"{row['pdb_id']}-{row['Chain']}-{row['Ligand']}-{row['Chain']}",
    }

    for _index, interest_row in ligands_df_samples.iterrows():
        current_row = deepcopy(protein_row)

        current_row["ligand_id"] = interest_row["ID"]
        current_row["smiles"] = interest_row["can_smiles"]
        current_row["ligand_category"] = interest_row["activity"]
        input_table_rows.append(deepcopy(current_row))

input_table = pd.DataFrame(input_table_rows)
input_table["is_pdb"] = 1
```
